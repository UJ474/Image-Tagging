This notebook explores the two visual feature representations evaluated across all Phase 1 Applied Machine Learning baselines:
1. **Flattened Raw Pixels ($784$ features):** Converts each normalized $28 \times 28$ grayscale apparel image directly into a 1D vector of length $784$.
2. **Histogram of Oriented Gradients (HOG) ($324$ features):** Captures local edge orientations using $9$ orientation bins, $7 \times 7$ pixel cells, $2 \times 2$ cell blocks, and L2-Hys normalization:
   $$\text{Total Features} = 3 \times 3 \text{ blocks} \times 2 \times 2 \text{ cells} \times 9 \text{ orientations} = 324$$

Below, we conduct an in-depth feature exploration:
- **Visualizing HOG Gradient Fields:** Side-by-side original image vs. HOG gradient reconstruction across all 10 garment categories.
- **Fine-Grained Ambiguity Analysis:** Comparing multiple samples of visually similar upper-body clothing (T-shirt/top, Pullover, Coat, Shirt) to illustrate why $7 \times 7$ cell pooling smooths away subtle collar and button details.
- **Orientation Profiles:** Profiling average gradient orientation distributions across classes to demonstrate structural silhouette signatures.
- **Geometric Information Tradeoff:** Analyzing dimensionality reduction, illumination invariance, and the motivation for Phase 2 deep learning.

```python
import sys
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
from skimage.feature import hog

# Locate project root and ensure src is importable
ROOT = next(
    (p for p in (Path.cwd(), *Path.cwd().parents)
     if (p / "pyproject.toml").is_file() and (p / "docs/team").is_dir()),
    None,
)
if ROOT is None:
    raise RuntimeError("Open this notebook from the repository root or notebooks/ directory.")
if str(ROOT / "src") not in sys.path:
    sys.path.insert(0, str(ROOT / "src"))

from aml.data import CLASS_NAMES, get_train_val_test_splits
from aml.features import extract_pixels, extract_hog_batch, extract_single_hog, get_features

plt.rcParams.update({"figure.dpi": 120, "axes.spines.top": False, "axes.spines.right": False})
figures_dir = ROOT / "reports" / "figures" / "features"
figures_dir.mkdir(parents=True, exist_ok=True)
print("Project root:", ROOT)
```

```python
# Load official development splits (normalized float32 in [0.0, 1.0])
data_dir = ROOT / "data" / "raw"
splits = get_train_val_test_splits(data_dir=data_dir)
X_train = splits["X_train"]
y_train = splits["y_train"]

print(f"Loaded X_train shape: {X_train.shape}, dtype: {X_train.dtype}, range: [{float(X_train.min()):.1f}, {float(X_train.max()):.1f}]")
print(f"Categories ({len(CLASS_NAMES)}): {CLASS_NAMES}")
```

## 1. Visualizing HOG Gradient Orientation Maps

For each of the 10 clothing categories, we select a representative sample from the training set, compute its HOG descriptors with `visualize=True`, and plot the original $28 \times 28$ image alongside its reconstructed gradient orientation field.

```python
representative_indices = [np.where(y_train == label_idx)[0][0] for label_idx in range(10)]

fig, axes = plt.subplots(10, 2, figsize=(6, 18), dpi=120)
fig.suptitle("Fashion-MNIST: Raw Images vs. HOG Gradient Maps", fontsize=13, y=0.995)

for row_idx, sample_idx in enumerate(representative_indices):
    label_idx = y_train[sample_idx]
    class_name = CLASS_NAMES[label_idx]
    img = X_train[sample_idx]
    
    # Compute HOG feature vector and visualization image
    _, hog_img = hog(
        img,
        orientations=9,
        pixels_per_cell=(7, 7),
        cells_per_block=(2, 2),
        block_norm="L2-Hys",
        transform_sqrt=False,
        visualize=True,
        feature_vector=True,
    )
    
    # Plot original grayscale image
    axes[row_idx, 0].imshow(img, cmap="gray", vmin=0.0, vmax=1.0)
    axes[row_idx, 0].set_title(f"Class {label_idx}: {class_name}", fontsize=9, loc="left", fontweight="bold")
    axes[row_idx, 0].axis("off")
    
    # Plot HOG gradient orientation map
    axes[row_idx, 1].imshow(hog_img, cmap="inferno")
    axes[row_idx, 1].set_title("HOG Gradient Map (324-d)", fontsize=9, loc="left")
    axes[row_idx, 1].axis("off")

plt.tight_layout()
output_fig_path = figures_dir / "hog_vs_pixels.png"
plt.savefig(output_fig_path, bbox_inches="tight", dpi=150)
print(f"Saved visualization figure to: {output_fig_path}")
plt.show()
```

## 2. Fine-Grained Ambiguity Deep Dive: The Upper-Body Dilemma

A core hypothesis of this project is that fixed, handcrafted descriptors like HOG succeed on distinct silhouettes (Trouser vs. Bag vs. Sneaker) but struggle on fine-grained apparel categories:
- **T-shirt/top (0)**
- **Pullover (2)**
- **Coat (4)**
- **Shirt (6)**

Because HOG pools gradients into $7 \times 7$ pixel cells, high-frequency spatial details—such as button plackets, collar points, and lapel seams—are averaged together.
Below, we visualize 3 distinct samples of each upper-body class with their corresponding HOG reconstructions.

```python
upper_classes = [0, 2, 4, 6]
samples_per_class = 3

fig, axes = plt.subplots(len(upper_classes) * 2, samples_per_class, figsize=(8, 14), dpi=120)
fig.suptitle("Upper-Body Clothing Ambiguity: 7x7 HOG Gradient Smoothing", fontsize=11, fontweight="bold", y=0.995)

for c_idx, c in enumerate(upper_classes):
    class_indices = np.where(y_train == c)[0][:samples_per_class]
    for s_idx, idx in enumerate(class_indices):
        img = X_train[idx]
        _, hog_img = hog(
            img, orientations=9, pixels_per_cell=(7, 7),
            cells_per_block=(2, 2), block_norm="L2-Hys",
            visualize=True, feature_vector=True
        )
        
        # Original image row
        row_orig = c_idx * 2
        axes[row_orig, s_idx].imshow(img, cmap="gray", vmin=0.0, vmax=1.0)
        axes[row_orig, s_idx].axis("off")
        if s_idx == 0:
            axes[row_orig, s_idx].set_title(f"{CLASS_NAMES[c]} (Original)", fontsize=9, loc="left", fontweight="bold")
            
        # HOG gradient row
        row_hog = c_idx * 2 + 1
        axes[row_hog, s_idx].imshow(hog_img, cmap="inferno")
        axes[row_hog, s_idx].axis("off")
        if s_idx == 0:
            axes[row_hog, s_idx].set_title(f"{CLASS_NAMES[c]} (HOG 324-d)", fontsize=8.5, loc="left", color="#333333")

plt.tight_layout()
upper_fig_path = figures_dir / "upper_body_hog_comparison.png"
plt.savefig(upper_fig_path, bbox_inches="tight", dpi=150)
print(f"Saved upper-body HOG comparison to: {upper_fig_path}")
plt.show()
```

## 3. Class Orientation Profiles: Gradient Magnitude across Orientation Bins

HOG partitions the gradient orientations into $9$ unsigned bins spanning $[0^\circ, 180^\circ)$ (each bin covering $20^\circ$).
We compute the mean HOG descriptor across $1,000$ samples per category and reshape it to measure energy distribution across orientation angles:
- **Trouser (1):** High concentration in vertical orientations ($80^\circ - 100^\circ$) due to parallel leg seams.
- **Sandal (5):** High concentration in horizontal orientations ($0^\circ - 20^\circ, 160^\circ - 180^\circ$) due to flat sole edges and horizontal straps.
- **T-shirt/top (0) & Shirt (6):** Diffuse, isotropic gradient distributions without a dominant single orientation, explaining their visual overlap.

```python
# Sample 500 images per class and calculate average orientation energy
n_eval_samples = 500
hog_by_class = {}

for c in range(10):
    c_imgs = X_train[y_train == c][:n_eval_samples]
    c_hogs = extract_hog_batch(c_imgs)
    # Reshape (N, 36 cells * 9 orientations) -> (N, 36, 9)
    c_hogs_reshaped = c_hogs.reshape(n_eval_samples, 36, 9)
    # Average magnitude per orientation bin
    hog_by_class[c] = np.mean(c_hogs_reshaped, axis=(0, 1))

orientation_angles = [f"{i*20}°" for i in range(9)]

fig, axes = plt.subplots(2, 5, figsize=(15, 6), dpi=120, sharey=True)
fig.suptitle("HOG Gradient Energy across 9 Orientation Bins [0° - 180°)", fontsize=12, fontweight="bold", y=0.99)

for c in range(10):
    ax = axes[c // 5, c % 5]
    bars = ax.bar(range(9), hog_by_class[c], color="#2a6f97", edgecolor="#012a4a", width=0.7)
    
    # Highlight highest orientation bin
    max_bin = np.argmax(hog_by_class[c])
    bars[max_bin].set_color("#e63946")
    
    ax.set_xticks(range(9))
    ax.set_xticklabels(orientation_angles, rotation=45, fontsize=7.5)
    ax.set_title(f"{c}: {CLASS_NAMES[c]}", fontsize=9, fontweight="bold")
    ax.grid(axis="y", linestyle="--", alpha=0.4)
    if c % 5 == 0:
        ax.set_ylabel("Mean Gradient Energy")

plt.tight_layout()
orient_fig_path = figures_dir / "hog_orientation_profiles.png"
plt.savefig(orient_fig_path, bbox_inches="tight", dpi=150)
print(f"Saved orientation profiles to: {orient_fig_path}")
plt.show()
```

## 4. Analysis of the Geometric Information Tradeoff

Comparing **Flattened Pixels (784 dimensions)** against **HOG Descriptors (324 dimensions)** highlights core considerations in classical feature engineering for apparel classification:

### 1. Spatial Structure vs. Contrast Invariance
- **Flattened Pixels ($784$ dimensions):** Retains raw grayscale luminance values and subtle internal surface textures, but flattening images into 1D vectors destroys 2D spatial adjacency. Classifiers like Logistic Regression or Linear SVM must independently learn weights for each pixel index without intrinsic awareness of neighboring pixels.
- **HOG Descriptors ($324$ dimensions):** Encodes localized 2D edge orientations within $7 \times 7$ pixel cells and applies L2-Hys block normalization over $2 \times 2$ cells. This introduces invariance to uniform illumination shifts and highlights primary garment silhouettes (e.g., trouser legs, bag straps, boot heels).

### 2. Dimensionality & Compression
- HOG reduces feature dimensionality from $784$ to $324$ (a **$58.7\%$ reduction**), speeding up downstream distance calculations for $k$-NN and reducing parameter count for linear models, while capturing structural edge envelopes.

### 3. The Clothing Ambiguity Gap (Phase 2 Motivation)
- Handcrafted HOG gradients effectively discriminate between classes with distinct structural silhouettes (such as **Trouser (1)** vs. **Bag (8)** vs. **Ankle boot (9)**).
- However, for visually similar tops—**T-shirt/top (0)**, **Pullover (2)**, **Coat (4)**, and **Shirt (6)**—outer garment silhouettes are almost identical. Crucial discriminative cues (collar geometry, button plackets, and sleeve seams) are averaged away within $7 \times 7$ cells.
- This fundamental bottleneck motivates **Phase 2**, where a Convolutional Neural Network (CNN) learns hierarchical, multi-scale visual filters directly from data rather than relying on fixed hand-engineered descriptors.

```python
# Verification of extraction dimensions and data types
test_samples = X_train[:5]

pixels_feat = extract_pixels(test_samples)
hog_feat = extract_hog_batch(test_samples)

print("--- Output Shape and Type Verification ---")
print(f"Input shape:            {test_samples.shape}, dtype: {test_samples.dtype}")
print(f"Flattened Pixels shape: {pixels_feat.shape}, dtype: {pixels_feat.dtype}")
print(f"HOG Descriptors shape:  {hog_feat.shape}, dtype: {hog_feat.dtype}")

assert pixels_feat.shape == (5, 784), f"Expected (5, 784), got {pixels_feat.shape}"
assert hog_feat.shape == (5, 324), f"Expected (5, 324), got {hog_feat.shape}"
assert pixels_feat.dtype == np.float32, f"Expected float32, got {pixels_feat.dtype}"
assert hog_feat.dtype == np.float32, f"Expected float32, got {hog_feat.dtype}"
assert np.all(np.isfinite(pixels_feat)), "Pixels contains non-finite values"
assert np.all(np.isfinite(hog_feat)), "HOG contains non-finite values"

print("All assertions passed successfully.")
```
